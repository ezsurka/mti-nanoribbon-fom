```python
import matplotlib.pyplot as plt

import numpy as np
from matplotlib.colors import ListedColormap, BoundaryNorm, to_rgba

white   = (1, 1, 1, 1)  # opaque
blue_a  = to_rgba('tab:blue',   alpha=0.2)
orange_a= to_rgba('tab:orange', alpha=0.2)

cmap = ListedColormap([white, blue_a, orange_a], name="white_blue20_orange20")
norm = BoundaryNorm([-0.5, 0.5, 1.5, 2.5], cmap.N)  # bins for values 0,1,2
```

```python
m0s = np.linspace(-80,80,1001)
Mzs = np.linspace(-80,80,1001)
```

```python
is_kap_u = np.reshape([(-(m0+Mz)<0).astype('uint8') for m0 in m0s for Mz in Mzs],(1001,1001))
is_kap_l = np.reshape([(-(m0-Mz)<0).astype('uint8') for m0 in m0s for Mz in Mzs],(1001,1001))
```

```python
m0_plot, Mz_plot = np.meshgrid(m0s, Mzs)
```

```python
plt.style.use(['default', './paper_thesis.mplstyle'])
```

```python
fig, ax = plt.subplots(1,1,figsize=(2.1,2.1))
ax.contourf(m0_plot,Mz_plot,is_kap_u+is_kap_l,cmap=cmap,linewidth=0)
#plt.colorbar(label="Nu. of edge states")
ax.set_xlabel("$M_z$ [meV]")
ax.set_ylabel("$m_0$ [meV]")
ax.text(-.25*80,.55*80,"QSHI")
ax.text(.35*80,-.05*80,"QAHI")
ax.text(-.85*80,-.05*80,"QAHI")
ax.text(-.1*80,-.65*80,"NI")
ax.set_xticks(np.arange(-80,100,40));
ax.set_yticks(np.arange(-80,100,40));
ax.set_aspect(1)

ax.arrow(0,0,0,20,width=1,length_includes_head=True);
ax.arrow(0,0,0,-20,width=1,length_includes_head=True);
ax.arrow(1,-20,40,0,width=1,length_includes_head=True);
ax.arrow(1,20,40,0,width=1,length_includes_head=True);

plt.savefig("num_edge_states.png")
plt.savefig("num_edge_states.pdf",bbox_inches="tight")
```
